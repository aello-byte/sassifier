```python
#env set-up
import pandas as pd
import spacy

nlp = spacy.load("en_core_web_sm")
pd.set_option("display.max_colwidth", 200)
```

```python

df = pd.read_parquet("../data/raw/tscript-0000.parquet")
df.head()
```

```python
non_empty_df = df[df['transcript'].astype(bool)]
```

```python
non_empty_df.head(10)
```

Converting transcript blob into sentences

```python
if "sentencizer" not in nlp.pipe_names:
    nlp.add_pipe("sentencizer")

sentence_rows = []

for idx, row in non_empty_df.iterrows():
    video_id = row["video_id"]
    url = row["original_url"]
    full_text = row["transcript"]

    doc = nlp(full_text)
    for sent in doc.sents:
        cleaned = sent.text.strip()
        if cleaned:
            sentence_rows.append({
                "video_id": video_id,
                "original_url": url,
                "transcript": cleaned
            })


clean_sent_df = pd.DataFrame(sentence_rows)
clean_sent_df = clean_sent_df[clean_sent_df['transcript'].str.len().between(30, 300)]

clean_sent_df.head()
```
